# Python Conditional Statement

```python
# Example of Python Conditional Statements
temp = 25

if temp < 15:
    print("It's cold outside!")
elif temp >= 15 and temp < 30:
    print("The weather is pleasant.")
else:
    print("It's hot outside!")
```

# Python Sets

```python
# Creating a Python Set
my_set = {1, 2, 3, 4, 4, 5}  # Duplicates are automatically removed
print("Initial Set:", my_set)

# Adding elements
my_set.add(6)
print("After adding 6:", my_set)

# Removing elements
my_set.remove(3)  # Raises KeyError if not present
my_set.discard(10)  # Safe remove, does not raise error if not present
print("After removing 3 and discarding 10:", my_set)

# Set Operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print("\nSet Operations with Set A {1,2,3,4} and Set B {3,4,5,6}:")
print("Union (A | B):", set_a.union(set_b))
print("Intersection (A & B):", set_a.intersection(set_b))
print("Difference (A - B):", set_a.difference(set_b))
print("Symmetric Difference (A ^ B):", set_a.symmetric_difference(set_b))
```

```python
# Checking if an element exists in a set
my_set = {'apple', 'banana', 'cherry'}

# Using 'in' operator
if 'banana' in my_set:
    print("'banana' is in the set!")
else:
    print("'banana' is NOT in the set!")

# Using 'not in' operator
if 'orange' not in my_set:
    print("'orange' is NOT in the set!")
```

```python
# Iterating through set elements using a loop
fruits = {'apple', 'banana', 'cherry'}

print("Iterating over the set:")
for fruit in fruits:
    print(fruit)
```

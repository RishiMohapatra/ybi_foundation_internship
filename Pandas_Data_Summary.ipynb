# Pandas Data Summary

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [24, 27, 22, 32, 29],
    'Salary': [50000, 54000, np.nan, 70000, 62000],
    'Department': ['HR', 'Engineering', 'Marketing', 'Engineering', 'HR']
}
df = pd.DataFrame(data)

print("--- DataFrame Info ---")
df.info()

print("\n--- Descriptive Statistics (Numeric) ---")
display(df.describe())

print("\n--- Descriptive Statistics (Categorical) ---")
display(df.describe(include=['O']))

print("\n--- Missing Values Count ---")
print(df.isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate average salary by department
avg_salary = df.groupby('Department')['Salary'].mean().reset_index()

# Create the bar chart
plt.figure(figsize=(8, 5))
sns.barplot(x='Department', y='Salary', data=avg_salary, palette='viridis', hue='Department', legend=False)
plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Average Salary ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot of Age vs Salary, colored by Department
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Age', y='Salary', hue='Department', style='Department', s=100)

plt.title('Age vs Salary by Department')
plt.xlabel('Age')
plt.ylabel('Salary ($)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Department')
plt.tight_layout()
plt.show()
```
